```python
def inspect(dia):
    if dia < 10:
        return "undersize"
    elif dia <= 20:
        return "in_spec"
    else:
        return "oversize"


def tally(labels):
    counts = {}
    for label in labels:
        counts[label] = counts.get(label, 0) + 1
    return counts


def report(shift, counts):
    return f"{shift}: {counts}"


labels = [inspect(12), inspect(8), inspect(22)]

counts = tally(labels)

print(report("Day shift", counts))
```

```python
# Function 1: Inspect diameter
def inspect(dia):
    """Check diameter and return its quality category."""
    if 24.95 <= dia <= 25.05:
        return "ACCEPT"
    elif 24.90 <= dia <= 25.10:
        return "REWORK"
    else:
        return "SCRAP"


# Function 2: Count categories
def tally(labels):
    """Count accepted, rework and scrap parts."""
    counts = {
        "ACCEPT": 0,
        "REWORK": 0,
        "SCRAP": 0
    }

    for label in labels:
        counts[label] += 1

    return counts


# Function 3: Generate report
def report(shift, counts):
    """Return a formatted shift report."""
    return (
        shift + " SHIFT REPORT: " +
        str(counts["ACCEPT"]) + " accepted, " +
        str(counts["REWORK"]) + " rework, " +
        str(counts["SCRAP"]) + " scrap"
    )


# Main program
log_lines = [
    "PE-26-0101 ; 25.02 ; DAY",
    "PE-26-0102 ; 24.97 ; DAY",
    "PE-26-0103 ; 25.15 ; DAY",
    "PE-26-0104 ; 24.99 ; NIGHT",
    "PE-26-0105 ; 25.01 ; DAY",
    "PE-26-0106 ; 24.85 ; NIGHT",
    "PE-26-0107 ; 25.08 ; DAY",
    "PE-26-0108 ; 25.00 ; NIGHT"
]

day_labels = []
night_labels = []

for line in log_lines:
    fields = line.strip().split(";")
    dia = float(fields[1].strip())
    shift = fields[2].strip().upper()

    label = inspect(dia)

    if shift == "DAY":
        day_labels.append(label)
    else:
        night_labels.append(label)

day_counts = tally(day_labels)
night_counts = tally(night_labels)

print(report("DAY", day_counts))
print(report("NIGHT", night_counts))
```

```python
def lookup(catalogue, code):
    return catalogue.get(code, "Not Found")


def scrap_codes(catalogue):
    scrap = set()

    for code, dia in catalogue.items():
        if dia < 10 or dia > 20:
            scrap.add(code)

    return scrap


catalogue = {
    "P101": 12,
    "P102": 8,
    "P103": 22,
    "P104": 15
}

print(lookup(catalogue, "P101"))
print(scrap_codes(catalogue))
```
